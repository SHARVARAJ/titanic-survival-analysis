#### Data Manipulation(of Embarked):

```python
Embarked = {'C':1,'Q':2,'S':3}
df['Embarked'] = df['Embarked'].map(Embarked)
df
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mean())
df['Embarked']
```

```python
#Checks if there is any NULL values
df.isnull().sum()
```

### Important 

```python
# To find different Age Groups:
def age_group(age):
    if age < 18:
        return "Child"
    elif age < 60:
        return "Adult"
    else:
        return "Senior"

df["Age_Group"] = df["Age"].apply(age_group)
df
```

```python
# To find Different Fare:
def fare_type(fare):
    if fare < 20:
        return "Low"
    elif fare < 50:
        return "Medium"
    else:
        return "High"

df["Fare_Type"] = df["Fare"].apply(fare_type)
df
```

```python
# To find Different Embarked:
def new_port(port):
    if port == 3.0:
        return "Southampton"
    elif port == 1.0:
        return "Cherbourg"
    elif port == 2.0:
        return "Queenstown"


df["Embarked"] = df["Embarked"].apply(new_port)
df
```

### Dashboard: Consolidated Survival Analysis (Matplotlib)

```python
import matplotlib.gridspec as gridspec

plt.rcParams['font.family'] = 'DejaVu Sans'

NAVY = "#1F3864"
GOLD = "#C9A227"
TEAL = "#2E7D6B"
CORAL = "#C15A4A"
BG = "#F7F7F5"
GRID = "#DDDDDD"

fig = plt.figure(figsize=(16, 10), facecolor=BG)
fig.suptitle("Titanic Survival Analysis \u2014 Exploratory Dashboard", fontsize=22, fontweight='bold', color=NAVY, y=0.985)
fig.text(0.5, 0.955, "Data cleaning, feature engineering & exploratory analysis  |  891 passengers  |  Python, Pandas, Matplotlib",
         ha='center', fontsize=11, color="#555555")

gs = gridspec.GridSpec(3, 3, figure=fig, hspace=0.55, wspace=0.35, top=0.90, bottom=0.06, left=0.06, right=0.97)

def style_ax(ax, title):
    ax.set_facecolor(BG)
    ax.set_title(title, fontsize=12, fontweight='bold', color=NAVY, pad=10)
    ax.grid(axis='y', color=GRID, linewidth=0.8, zorder=0)
    ax.set_axisbelow(True)
    for spine in ['top', 'right']:
        ax.spines[spine].set_visible(False)
    ax.spines['left'].set_color('#999999')
    ax.spines['bottom'].set_color('#999999')

# 1. Overall survival rate (donut)
ax1 = fig.add_subplot(gs[0, 0])
survived = df['Survived'].sum()
died = len(df) - survived
ax1.pie([survived, died], labels=['Survived', 'Did not survive'], colors=[TEAL, CORAL],
        autopct='%1.1f%%', startangle=90, wedgeprops=dict(width=0.42, edgecolor=BG, linewidth=2),
        textprops={'fontsize': 9, 'color': '#333333'})
ax1.set_title("Overall Survival Rate", fontsize=12, fontweight='bold', color=NAVY, pad=10)

# 2. Survival rate by gender
ax2 = fig.add_subplot(gs[0, 1])
g = df.groupby('Sex')['Survived'].mean() * 100
bars = ax2.bar(g.index.str.capitalize(), g.values, color=[CORAL, TEAL], width=0.55, zorder=3)
style_ax(ax2, "Survival Rate by Gender")
ax2.set_ylabel("Survival %", fontsize=9)
for b in bars:
    ax2.text(b.get_x()+b.get_width()/2, b.get_height()+1.5, f"{b.get_height():.0f}%", ha='center', fontsize=9, color=NAVY, fontweight='bold')

# 3. Survival rate by class
ax3 = fig.add_subplot(gs[0, 2])
c = df.groupby('Pclass')['Survived'].mean() * 100
bars = ax3.bar(['1st', '2nd', '3rd'], c.values, color=GOLD, width=0.55, zorder=3)
style_ax(ax3, "Survival Rate by Class")
ax3.set_ylabel("Survival %", fontsize=9)
for b in bars:
    ax3.text(b.get_x()+b.get_width()/2, b.get_height()+1.5, f"{b.get_height():.0f}%", ha='center', fontsize=9, color=NAVY, fontweight='bold')

# 4. Age distribution
ax4 = fig.add_subplot(gs[1, 0])
ax4.hist(df['Age'], bins=20, color=NAVY, edgecolor=BG, zorder=3)
style_ax(ax4, "Age Distribution")
ax4.set_xlabel("Age", fontsize=9)
ax4.set_ylabel("Passengers", fontsize=9)

# 5. Passenger class distribution
ax5 = fig.add_subplot(gs[1, 1])
pc = df['Pclass'].value_counts().sort_index()
ax5.bar(['1st', '2nd', '3rd'], pc.values, color=TEAL, width=0.55, zorder=3)
style_ax(ax5, "Passenger Count by Class")
ax5.set_ylabel("Passengers", fontsize=9)

# 6. Embarkation port distribution
ax6 = fig.add_subplot(gs[1, 2])
ep = df['Embarked'].value_counts()
ax6.bar(ep.index, ep.values, color=CORAL, width=0.55, zorder=3)
style_ax(ax6, "Passengers by Embarkation Port")
ax6.set_ylabel("Passengers", fontsize=9)
ax6.tick_params(axis='x', labelrotation=10)

# 7. Average fare by class
ax7 = fig.add_subplot(gs[2, 0])
fc = df.groupby('Pclass')['Fare'].mean()
ax7.bar(['1st', '2nd', '3rd'], fc.values, color=GOLD, width=0.55, zorder=3)
style_ax(ax7, "Average Fare by Class")
ax7.set_ylabel("Fare ($)", fontsize=9)

# 8. Survival rate by age group
ax8 = fig.add_subplot(gs[2, 1])
ag = df.groupby('Age_Group')['Survived'].mean().reindex(['Child','Adult','Senior']) * 100
bars = ax8.bar(ag.index, ag.values, color=NAVY, width=0.55, zorder=3)
style_ax(ax8, "Survival Rate by Age Group")
ax8.set_ylabel("Survival %", fontsize=9)
for b in bars:
    ax8.text(b.get_x()+b.get_width()/2, b.get_height()+1.5, f"{b.get_height():.0f}%", ha='center', fontsize=9, color=NAVY, fontweight='bold')

# 9. Survival rate by embarkation port
ax9 = fig.add_subplot(gs[2, 2])
epsurv = df.groupby('Embarked')['Survived'].mean() * 100
bars = ax9.bar(epsurv.index, epsurv.values, color=TEAL, width=0.55, zorder=3)
style_ax(ax9, "Survival Rate by Port")
ax9.set_ylabel("Survival %", fontsize=9)
ax9.tick_params(axis='x', labelrotation=10)
for b in bars:
    ax9.text(b.get_x()+b.get_width()/2, b.get_height()+1.5, f"{b.get_height():.0f}%", ha='center', fontsize=9, color=NAVY, fontweight='bold')

plt.savefig("titanic_dashboard.png", dpi=150, facecolor=BG, bbox_inches='tight')
plt.show()
```
